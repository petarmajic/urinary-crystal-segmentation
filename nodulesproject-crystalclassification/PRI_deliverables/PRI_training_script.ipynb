```python
import os, glob, yaml
from pathlib import Path
from collections import Counter

ds_root = Path("/content/ds/cristals-YOLO_Ultralytics_v1.0")

# load yaml
data = yaml.safe_load((ds_root / "data.yaml").read_text(encoding="utf-8", errors="ignore"))
names = data["names"]
nc = max(int(k) for k in names.keys()) + 1
class_names = [None] * nc
for k, v in names.items():
    class_names[int(k)] = v

# --- merge mapping: old_id -> canonical_name ---
merge_map = {}

# default: keep original
for i in range(nc):
    merge_map[i] = class_names[i]

# merge duplicates INTO the coded version
merge_map[0] = class_names[13]   # brushite -> Br_brushite
merge_map[1] = class_names[17]   # calcite  -> CAL_calcite
merge_map[2] = class_names[16]   # caoxite  -> C3_caoxite
merge_map[3] = class_names[22]   # cystine  -> CYS_cystine
merge_map[4] = class_names[23]   # dihydroxyadenine -> DHAd_dihydroxyadénine

# build new class list (unique)
new_class_names = sorted(set(merge_map.values()))
new_id_by_name = {name: idx for idx, name in enumerate(new_class_names)}

print("Old nc:", nc)
print("New nc after merge:", len(new_class_names))

# parse class id (standard yolo-seg)
def class_id_from_line(line: str):
    t = line.strip().split()
    if not t:
        return None
    try:
        return int(float(t[0]))
    except:
        return None

# collect matched train labeled images
img_files = [p for p in glob.glob(str(ds_root / "images" / "train" / "**" / "*"), recursive=True) if os.path.isfile(p)]
lbl_files = [p for p in glob.glob(str(ds_root / "labels" / "train" / "**" / "*.txt"), recursive=True) if os.path.isfile(p)]

img_names = {Path(p).stem for p in img_files}
lbl_by_name = {Path(p).stem: Path(p) for p in lbl_files}
matched = sorted(img_names & set(lbl_by_name.keys()))

images_per_class = Counter()
instances_per_class = Counter()

for name in matched:
    lp = lbl_by_name[name]
    lines = lp.read_text(encoding="utf-8", errors="ignore").splitlines()

    present = set()
    for ln in lines:
        ln = ln.strip()
        if not ln:
            continue

        old_id = class_id_from_line(ln)
        if old_id is None or old_id < 0 or old_id >= nc:
            continue

        new_name = merge_map[old_id]
        new_id = new_id_by_name[new_name]

        instances_per_class[new_id] += 1
        present.add(new_id)

    for c in present:
        images_per_class[c] += 1

print("\nMatched labeled images:", len(matched))
print(f"\n{'new_id':>6}  {'class':<45}  {'images':>6}  {'instances':>9}")
print("-"*75)

for cid in range(len(new_class_names)):
    print(f"{cid:>6}  {new_class_names[cid]:<45}  {images_per_class.get(cid,0):>6}  {instances_per_class.get(cid,0):>9}")
```

```python
kept_ids = [cid for cid in range(len(new_class_names)) if images_per_class.get(cid, 0) > 0]
kept_class_names = [new_class_names[cid] for cid in kept_ids]

print("Kept classes:", len(kept_class_names))
print("Dropped (zero) classes:", len(new_class_names) - len(kept_class_names))

dropped = [new_class_names[cid] for cid in range(len(new_class_names)) if cid not in kept_ids]
print("\nDropped classes (images=0):")
for name in dropped:
    print(" -", name)

# 2) map merged_id -> final_id
merged_to_final = {merged_id: final_id for final_id, merged_id in enumerate(kept_ids)}

# 3) build old_id -> final_id
# old_id -> merged_name -> merged_id -> final_id
new_id_by_name = {name: idx for idx, name in enumerate(new_class_names)}
old_to_final = {}
for old_id, merged_name in merge_map.items():
    merged_id = new_id_by_name[merged_name]
    if merged_id in merged_to_final:
        old_to_final[old_id] = merged_to_final[merged_id]
```

```python
import os, shutil, yaml
from pathlib import Path

clean_root = ds_root / "dataset_clean"
img_out = clean_root / "images" / "all"
lbl_out = clean_root / "labels" / "all"
img_out.mkdir(parents=True, exist_ok=True)
lbl_out.mkdir(parents=True, exist_ok=True)

# final mapping: merged_id -> final_id (0..)
final_class_list = sorted(list(kept_merged_ids))
merged_to_final = {merged_id: i for i, merged_id in enumerate(final_class_list)}
final_names = [new_class_names[mid] for mid in final_class_list]

# helper: find image path for a basename (across images/train/**)
# (ovo koristi img_by_name i lbl_by_name iz ranijih koraka; ako nemaš img_by_name, izgradimo ga)
img_files_all = [Path(p) for p in glob.glob(str(ds_root / "images" / "**" / "*"), recursive=True) if os.path.isfile(p)]
img_by_name = {p.stem: p for p in img_files_all}

print("Creating clean dataset at:", clean_root)
print("Final classes:", len(final_names))
print("Usable images:", len(usable_images))

def remap_label_file(old_label_path: Path) -> str:
    """Return remapped label content as text (keeps only kept classes, remaps ids)."""
    out_lines = []
    lines = old_label_path.read_text(encoding="utf-8", errors="ignore").splitlines()

    for ln in lines:
        ln = ln.strip()
        if not ln:
            continue

        t = ln.split()
        old_id = int(float(t[0]))          # YOLO-seg: first token is class_id
        merged_name = merge_map[old_id]    # old -> merged name
        merged_id = new_id_by_name[merged_name]

        if merged_id not in kept_merged_ids:
            continue

        new_id = merged_to_final[merged_id]
        t[0] = str(new_id)                # replace class id
        out_lines.append(" ".join(t))

    return "\n".join(out_lines) + ("\n" if out_lines else "")

# copy + write
written = 0
skipped_empty = 0

for name in usable_images:
    src_img = img_by_name[name]
    src_lbl = lbl_by_name[name]

    dst_img = img_out / src_img.name      # keep filename + extension
    dst_lbl = lbl_out / (name + ".txt")

    shutil.copy2(src_img, dst_img)

    new_lbl_text = remap_label_file(src_lbl)
    if new_lbl_text.strip() == "":
        skipped_empty += 1
        continue

    dst_lbl.write_text(new_lbl_text, encoding="utf-8")
    written += 1

print("Copied images:", len(usable_images))
print("Written labels:", written)
print("Skipped empty labels after filter:", skipped_empty)

# write new data.yaml
clean_yaml = {
    "path": str(clean_root),
    "train": "images/all",   # kasnije K-fold će koristiti txt liste
    "val": "images/all",
    "names": {i: n for i, n in enumerate(final_names)}
}
(clean_root / "data.yaml").write_text(yaml.safe_dump(clean_yaml, sort_keys=False, allow_unicode=True), encoding="utf-8")

print("Clean data.yaml written:", clean_root / "data.yaml")
```

```python
import numpy as np
from iterstrat.ml_stratifiers import MultilabelStratifiedKFold

clean_root = ds_root / "dataset_clean"
img_dir = clean_root / "images" / "all"
lbl_dir = clean_root / "labels" / "all"

# collect clean images + labels
img_paths = sorted([p for p in img_dir.iterdir() if p.is_file()])
lbl_by_stem = {p.stem: p for p in lbl_dir.iterdir() if p.is_file() and p.suffix == ".txt"}

# keep only those with labels
X = []
Y = []
nc_final = len(final_names)

for img in img_paths:
    stem = img.stem
    if stem not in lbl_by_stem:
        continue
    lbl = lbl_by_stem[stem]
    lines = lbl.read_text(encoding="utf-8", errors="ignore").splitlines()

    present = set()
    for ln in lines:
        ln = ln.strip()
        if not ln:
            continue
        c = int(float(ln.split()[0]))
        if 0 <= c < nc_final:
            present.add(c)

    if not present:
        continue

    # x_rel must be relative to the 'path' entry in data.yaml (which is clean_root)
    x_rel = str(Path("images/all") / img.name)  # Reverted to original logic for correct relative path
    y = np.zeros(nc_final, dtype=np.int8)
    for c in present:
        y[c] = 1

    X.append(x_rel)
    Y.append(y)

X = np.array(X)
Y = np.array(Y)

print("Clean usable images for kfold:", len(X))
print("Y shape:", Y.shape)

k = 5
mskf = MultilabelStratifiedKFold(n_splits=k, shuffle=True, random_state=42)

kfold_dir = clean_root / "kfold"
kfold_dir.mkdir(exist_ok=True)

for fold, (tr, va) in enumerate(mskf.split(np.zeros(len(X)), Y), start=1):
    train_txt_path = kfold_dir / f"fold_{fold}_train.txt"
    val_txt_path   = kfold_dir / f"fold_{fold}_val.txt"
    fold_yaml_path = kfold_dir / f"data_fold_{fold}.yaml"

    train_txt_path.write_text("\n".join(X[tr].tolist()) + "\n", encoding="utf-8")
    val_txt_path.write_text("\n".join(X[va].tolist()) + "\n", encoding="utf-8")

    yml = {
        "path": str(clean_root),
        "train": str(train_txt_path.relative_to(clean_root)),  # Make path relative to clean_root
        "val": str(val_txt_path.relative_to(clean_root)),      # Make path relative to clean_root
        "names": {i: n for i, n in enumerate(final_names)},
        "channels": 3 # Explicitly set channels to 3
    }
    fold_yaml_path.write_text(yaml.safe_dump(yml, sort_keys=False, allow_unicode=True), encoding="utf-8")

    print(f"Fold {fold}: train={len(tr)} val={len(va)} -> {fold_yaml_path.name}")

print("\nK-fold ready in:", kfold_dir)
!ls -lah "$kfold_dir"
```

```python
from pathlib import Path
import yaml

clean_root = Path("/content/ds/cristals-YOLO_Ultralytics_v1.0/dataset_clean")
kfold_dir  = clean_root / "kfold"

# učitaj names iz clean data.yaml
clean_data = yaml.safe_load((clean_root / "data.yaml").read_text(encoding="utf-8"))
names = clean_data["names"]

def make_abs_list(txt_in: Path, txt_out: Path):
    lines = [l.strip() for l in txt_in.read_text(encoding="utf-8").splitlines() if l.strip()]
    abs_lines = []
    for l in lines:
        p = Path(l)
        if not p.is_absolute():
            p = (clean_root / p)  # images/all/xxx -> /content/.../dataset_clean/images/all/xxx
        abs_lines.append(str(p))
    txt_out.write_text("\n".join(abs_lines) + "\n", encoding="utf-8")

for fold in range(1, 6):
    tr_in = kfold_dir / f"fold_{fold}_train.txt"
    va_in = kfold_dir / f"fold_{fold}_val.txt"

    tr_out = kfold_dir / f"fold_{fold}_train_abs.txt"
    va_out = kfold_dir / f"fold_{fold}_val_abs.txt"

    make_abs_list(tr_in, tr_out)
    make_abs_list(va_in, va_out)

    # novi yaml (path može biti bilo što kad su putanje apsolutne, ali stavimo clean_root radi reda)
    fold_yaml = {
        "path": str(clean_root),
        "train": str(tr_out),
        "val": str(va_out),
        "names": names,
        "channels": 3 # Explicitly set channels to 3 for the _abs.yaml files
    }
    (kfold_dir / f"data_fold_{fold}_abs.yaml").write_text(
        yaml.safe_dump(fold_yaml, sort_keys=False, allow_unicode=True),
        encoding="utf-8"
    )

print("Done. New files:")
!ls -lah "$kfold_dir" | sed -n '1,120p'
```

```python
from pathlib import Path
from PIL import Image

img_dir = Path("/content/ds/cristals-YOLO_Ultralytics_v1.0/dataset_clean/images/all")

bad = []
ok = 0
failed = 0

for p in sorted(img_dir.iterdir()):
    if not p.is_file():
        continue
    try:
        with Image.open(p) as im:
            bands = im.getbands()   # npr. ('R','G','B') ili ('R','G','B','A')
            mode = im.mode          # npr. RGB, RGBA, L, LA, CMYK...
            size = im.size
            ch = len(bands)

        if not (mode == "RGB" and ch == 3):
            bad.append((p.name, mode, ch, bands, size))
        else:
            ok += 1

    except Exception as e:
        failed += 1
        bad.append((p.name, "READ_ERROR", None, None, None))

print("Total files:", ok + len(bad))
print("OK RGB files:", ok)
print("Non-RGB or problematic files:", len(bad))
print("Read errors:", failed)

# ispiši sve (ili prvih 50 ako ih je puno)
for row in bad[:200]:
    print(row)
```

```python
from pathlib import Path
from collections import Counter
import shutil
import yaml
import numpy as np
from iterstrat.ml_stratifiers import MultilabelStratifiedKFold

# ---------------- CONFIG ----------------
src_root = Path("/content/ds/cristals-YOLO_Ultralytics_v1.0/dataset_clean")
dst_root = Path("/content/ds/cristals-YOLO_Ultralytics_v1.0/dataset_clean_drop4")

drop_names = {
    "ATZ_atazanavir",
    "Cylindre",
    "CEF_ceftriaxonate_de_calcium",
    "CHOL_cholestérol",
}
```

```python
k = 5
seed = 42

img_src = src_root / "images" / "all"
lbl_src = src_root / "labels" / "all"

img_dst = dst_root / "images" / "all"
lbl_dst = dst_root / "labels" / "all"
kfold_dir = dst_root / "kfold"

for d in [img_dst, lbl_dst, kfold_dir]:
    d.mkdir(parents=True, exist_ok=True)
```

```python
data_yaml_path = src_root / "data.yaml"
data = yaml.safe_load(data_yaml_path.read_text(encoding="utf-8"))

names = data["names"]
if isinstance(names, list):
    old_id_to_name = {i: n for i, n in enumerate(names)}
else:
    old_id_to_name = {int(k): v for k, v in names.items()}

# pronađi drop class_ids
drop_old_ids = {cid for cid, n in old_id_to_name.items() if n in drop_names}
keep_old_ids = [cid for cid in sorted(old_id_to_name.keys()) if cid not in drop_old_ids]

print("Dropping classes:")
for cid in sorted(drop_old_ids):
    print(f" - {cid}: {old_id_to_name[cid]}")
```

```python
old_to_new = {old_id: new_id for new_id, old_id in enumerate(keep_old_ids)}
new_names = {new_id: old_id_to_name[old_id] for old_id, new_id in old_to_new.items()}
```

```python
print("\nKept classes:", len(new_names))

# ---------------- 2) prođi kroz sve label fileove, filtriraj linije, kopiraj slike ----------------
# indeks slika po stemu
img_by_stem = {}
for p in img_src.iterdir():
    if p.is_file():
        img_by_stem[p.stem] = p

usable_stems = []
images_per_class = Counter()
instances_per_class = Counter()

for lp in sorted(lbl_src.glob("*.txt")):
    stem = lp.stem
    lines = lp.read_text(encoding="utf-8", errors="ignore").splitlines()

    kept_lines = []
    present = set()

    for ln in lines:
        ln = ln.strip()
        if not ln:
            continue
        parts = ln.split()
        old_id = int(float(parts[0]))
        if old_id in drop_old_ids:
            continue
        if old_id not in old_to_new:
            continue
        new_id = old_to_new[old_id] # Corrected indentation
        parts[0] = str(new_id)
        kept_lines.append(" ".join(parts))

        instances_per_class[new_id] += 1
        present.add(new_id)

    if not kept_lines:
        continue  # slika postaje prazna -> drop

    # kopiraj label
    (lbl_dst / f"{stem}.txt").write_text("\n".join(kept_lines) + "\n", encoding="utf-8")
    # kopiraj sliku (ako postoji)
    ip = img_by_stem.get(stem)
    if ip is None:
        # ako fali slika, preskoči i pobriši label da ne ostane dangling
        (lbl_dst / f"{stem}.txt").unlink(missing_ok=True)
        continue

    shutil.copy2(ip, img_dst / ip.name)
    usable_stems.append(stem)

    for c in present:
        images_per_class[c] += 1

print("\nUsable images after drop:", len(usable_stems))
```

```python
dst_data = {
    "path": str(dst_root),
    "names": new_names
}
(dst_root / "data.yaml").write_text(
    yaml.safe_dump(dst_data, sort_keys=False, allow_unicode=True),
    encoding="utf-8"
)
```

```python
print(f"\n{'new_id':>6}  {'class':<45}  {'images':>6}  {'instances':>9}")
print("-"*72)
for cid in range(len(new_names)):
    print(f"{cid:>6}  {new_names[cid]:<45}  {images_per_class.get(cid,0):>6}  {instances_per_class.get(cid,0):>9}")

# ---------------- 5) multilabel Y za stratified kfold ----------------
# X = apsolutne putanje slika u novom folderu
X, Y = [], []
nC = len(new_names)

# map label -> prisutne klase (po slici)
for img in sorted(img_dst.iterdir()):
    if not img.is_file():
        continue
    lbl = lbl_dst / f"{img.stem}.txt"
    if not lbl.exists():
        continue

    present = set()
    for ln in lbl.read_text(encoding="utf-8", errors="ignore").splitlines():
        ln = ln.strip()
        if not ln:
            continue
        c = int(float(ln.split()[0]))
        if 0 <= c < nC:
            present.add(c)

    if not present:
        continue

    y = np.zeros(nC, dtype=np.int8)
    for c in present:
        y[c] = 1

    X.append(str(img))
    Y.append(y)

X = np.array(X)
Y = np.array(Y)

print("\nK-fold base shapes:", X.shape, Y.shape)

```

```python
mskf = MultilabelStratifiedKFold(n_splits=k, shuffle=True, random_state=seed)

for fold, (tr, va) in enumerate(mskf.split(np.zeros(len(X)), Y), start=1):
    train_txt = kfold_dir / f"fold_{fold}_train_abs.txt"
    val_txt   = kfold_dir / f"fold_{fold}_val_abs.txt"
    yaml_out  = kfold_dir / f"data_fold_{fold}_abs.yaml"

    train_txt.write_text("\n".join(X[tr].tolist()) + "\n", encoding="utf-8")
    val_txt.write_text("\n".join(X[va].tolist()) + "\n", encoding="utf-8")

    fold_yaml = {
        "path": str(dst_root),
        "train": str(train_txt),
        "val": str(val_txt),
        "names": new_names
    }
    yaml_out.write_text(yaml.safe_dump(fold_yaml, sort_keys=False, allow_unicode=True), encoding="utf-8")

    print(f"Fold {fold}: train={len(tr)} val={len(va)} -> {yaml_out.name}")

print("\nNew dataset ready:", dst_root)
print("Kfold ready:", kfold_dir)
```
